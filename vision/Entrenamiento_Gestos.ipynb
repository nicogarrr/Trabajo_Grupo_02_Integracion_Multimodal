## 3. Crear el script de entrenamiento
Se ha creado un script en tu ordenador local llamado `entrenar_gestos.py` con las instrucciones de la práctica. Sube ese archivo al entorno de Colab. No hay necesidad de copiar el código aquí, simplemente sube el archivo .py que acabo de crear.

### Qué hace este script:
A. **Carga y Organización:** Renombra automáticamente tu carpeta "negativa" o "ninguno" a `None` (requisito de MediaPipe) y carga el dataset usando `gesture_recognizer.Dataset.from_folder()`.
B. **Split:** Divide en 80% entrenamiento, 10% validación y 10% test.
C. **HParams:** Configura `epochs=10` y `export_dir`.
D. **Ejecución y Exportación:** Entrena el modelo y lo guarda en la carpeta `modelo_exportado`.

```python
%%writefile entrenar_gestos.py
import os
import shutil
from mediapipe_model_maker import gesture_recognizer

# 1. Ajustar estructura de carpetas
base_path = "dataset"
export_path = "modelo_exportado"

# Si el zip se descomprimió con una subcarpeta extra (dataset/dataset/...)
if os.path.exists("dataset/dataset"):
    base_path = "dataset/dataset"

# LIMPIEZA: Eliminar exportaciones previas para evitar error de compatibilidad de checkpoints
if os.path.exists(export_path):
    print(f"Limpiando carpeta de exportación previa: {export_path}")
    shutil.rmtree(export_path)

print(f"Explorando carpetas en: {base_path}")
subfolders = [f for f in os.listdir(base_path) if os.path.isdir(os.path.join(base_path, f))]
print(f"Carpetas encontradas: {subfolders}")

# Renombrar posibles carpetas negativas a 'None'
posibles_negativos = ['negativas', 'ninguno', 'background', 'negative', 'nada', 'Ninguno', 'Negativas']
found_none = 'None' in subfolders

if not found_none:
    for folder in posibles_negativos:
        if folder in subfolders:
            os.rename(os.path.join(base_path, folder), os.path.join(base_path, 'None'))
            print(f"Carpeta '{folder}' renombrada a 'None'.")
            found_none = True
            break

if not found_none:
    print("ADVERTENCIA: ADVERTENCIA: No se encontró una carpeta de negativos. Creando carpeta 'None' vacía para evitar error.")
    os.makedirs(os.path.join(base_path, 'None'), exist_ok=True)

# 2. Carga de Datos
print(f"Cargando dataset desde: {base_path}")
data = gesture_recognizer.Dataset.from_folder(
    dirname=base_path,
    hparams=gesture_recognizer.HandDataPreprocessingParams()
)

# 3. Split
train_data, rest_data = data.split(0.8)
validation_data, test_data = rest_data.split(0.5)

# 4. HParams
hparams = gesture_recognizer.HParams(
    export_dir=export_path,
    epochs=10,
    batch_size=2
)
options = gesture_recognizer.GestureRecognizerOptions(hparams=hparams)

# 5. Entrenamiento
print("\nIniciando entrenamiento...")
model = gesture_recognizer.GestureRecognizer.create(
    train_data=train_data,
    validation_data=validation_data,
    options=options
)

# 6. Evaluación
print("\nEvaluando...")
loss, acc = model.evaluate(test_data, batch_size=1)
print(f"Test loss:{loss}, Test accuracy:{acc}")

# 7. Exportación
model.export_model()
print(f"Modelo exportado en '{export_path}/gesture_recognizer.task'")
```
